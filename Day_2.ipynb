```python
#PART 1 (EVEN)


def id_check(s:str):
    count = 0
    rang = s.split('-')    
    low = int(rang[0])
    high = int(rang[1])


    for i in range(low,high+1):
        i = str(i)
        if len(i) % 2 == 0:
            mid = len(i) // 2
            fh = i[:mid]
            sh = i[mid:]
            if fh == sh:
                count += int(i)
        else:
            continue
    return count
```

```python
#Part 2 (ODDS N EVENS)

def id_check_xl(s:str):
    count = 0
    rang = s.split('-')    
    low = int(rang[0])
    high = int(rang[1])

    for i in range(low,high+1):
        i = str(i)
        long = len(i)

        #traditional (123123,11,22)
        if long % 2 == 0:
            mid = long // 2
            fh = i[:mid]
            sh = i[mid:]
            if fh == sh:
                count += int(i)
            
            #even repeat (121212)
            else:
                for j in range(1, long // 2 + 1):
                    if long % j == 0:
                        pat = i[:j]
                        if pat * (long // j) == i:
                            count += int(i)

        #odds (111, 856856856)
        else:
            for j in range(1, long // 2 + 1):
                if long % j == 0:
                    pat = s[:j]
                    if pat * (long // j) == i:
                        count += int(i)
    return count

#pub set
ex = ex.split(',')

#assigned set
test = test.split(',')

total = 0

for id in ex:
    total += id_check_xl(id)

print(total)  
```
